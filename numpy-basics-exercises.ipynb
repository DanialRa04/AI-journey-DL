# NumPy Basics Exercises

A short practice notebook built from my original `04-Numpy Exercises.ipynb` while reviewing core NumPy array operations.

## Goal

I wanted a compact notebook that checks the basics: creating arrays, reshaping data, sampling random values, slicing matrices, and applying simple aggregations.

````python
import numpy as np

np.random.seed(42)
````

## Array Creation

Create an array of 10 zeros.

````python
np.zeros(10)
````

Create an array of 10 ones.

````python
np.ones(10)
````

Create an array of 10 fives.

````python
np.full(10, 5)
````

Create an array of the integers from 10 to 50.

````python
np.arange(10, 51)
````

Create an array of all even integers from 10 to 50.

````python
np.arange(10, 51, 2)
````

Create a `3 x 3` matrix with values ranging from 0 to 8.

````python
np.arange(9).reshape(3, 3)
````

Create a `3 x 3` identity matrix.

````python
np.eye(3, dtype=int)
````

## Random Values and Reshaping

Generate one random number between 0 and 1.

````python
np.random.rand()
````

Generate 25 random numbers from a standard normal distribution.

````python
np.random.randn(25)
````

Build a `10 x 10` matrix with values from `0.01` to `1.00`.

````python
(np.arange(1, 101) / 100).reshape(10, 10)
````

Create 20 linearly spaced points between 0 and 1.

````python
np.linspace(0, 1, 20)
````

## Indexing and Selection

Use the matrix below for the next few selections.

````python
mat = np.arange(1, 26).reshape(5, 5)
mat
````

Select the block:

```python
array([[12, 13, 14, 15],
       [17, 18, 19, 20],
       [22, 23, 24, 25]])
```

````python
mat[2:, 1:]
````

Get the value `20`.

````python
mat[3, 4]
````

Select the column vector:

```python
array([[ 2],
       [ 7],
       [12]])
```

````python
mat[:3, 1:2]
````

Get the final row of the matrix.

````python
mat[4, :]
````

Get the bottom two rows.

````python
mat[3:, :]
````

## Aggregations

Get the sum of all values in `mat`.

````python
mat.sum()
````

Get the standard deviation of the values in `mat`.

````python
mat.std()
````

Get the sum of each column in `mat`.

````python
mat.sum(axis=0)
````
