# Africa Crisis Pandas Exercises

## Goal
Work through a small set of pandas exercises on African banking and systemic crisis history. I kept the original exercise spirit, but cleaned the structure and made the file paths repo-relative so the notebook runs as-is.

## Data
The notebook uses `data/african_econ_crises.csv`, a public dataset with yearly country-level records for systemic crises, inflation, exchange rates, and debt defaults.

```python
from pathlib import Path

import numpy as np
import pandas as pd

pd.set_option('display.max_columns', None)

DATA_PATH = Path('data') / 'african_econ_crises.csv'
df = pd.read_csv(DATA_PATH)
df.head()
```

## Quick Checks
Start with a fast inspection so the later answers have some context.

```python
summary = {
    'rows': len(df),
    'columns': len(df.columns),
    'year_range': (int(df['year'].min()), int(df['year'].max())),
    'country_count': int(df['country'].nunique()),
}
summary
```

## Country Coverage
These answers show which countries are present and how broad the dataset is.

```python
country_count = df['country'].nunique()
countries = sorted(df['country'].unique())

country_count, countries
```

## Inflation Outlier
Find the country-year pair with the most extreme annual CPI inflation in the dataset.

```python
highest_inflation = df.loc[df['inflation_annual_cpi'].idxmax(), ['country', 'year', 'inflation_annual_cpi']]
highest_inflation.to_frame().T
```

## Kenya's First Systemic Crisis
Filter the Kenya rows, keep crisis years only, and read the earliest one.

```python
kenya_first_crisis_year = int(
    df.loc[(df['country'] == 'Kenya') & (df['systemic_crisis'] == 1), 'year'].min()
)
kenya_first_crisis_year
```

## Systemic Crises By Country
Summing the `systemic_crisis` flag gives the number of crisis years recorded for each country.

```python
systemic_crises_per_country = (
    df.groupby('country', as_index=True)['systemic_crisis']
      .sum()
      .sort_values(ascending=False)
)
systemic_crises_per_country
```

## Debt Default And Exchange Rate Questions
Finish the exercise set with one debt-default lookup and one exchange-rate lookup.

```python
zimbabwe_external_defaults = int(
    df.loc[
        (df['country'] == 'Zimbabwe') & (df['sovereign_external_debt_default'] == 1),
        'year'
    ].count()
)

algeria_highest_exchange_year = df.loc[
    df.loc[df['country'] == 'Algeria', 'exch_usd'].idxmax(),
    ['country', 'year', 'exch_usd']
]

zimbabwe_external_defaults, algeria_highest_exchange_year.to_dict()
```

## Result
The exercise confirms a few good pandas habits: use `nunique()` and `unique()` for quick categorical checks, `idxmax()` to pull the full row behind an extreme value, boolean masks for targeted lookups, and `groupby()` for country-level summaries.
