# 3D Plotting Basics

## Goal
Build a simple synthetic curve with NumPy, then compare how it looks in 3D and in a flat 2D view.

## Data
There is no external dataset here. The notebook creates its own coordinates with `numpy.linspace`, `sin`, and `cos`.


```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits import mplot3d

plt.style.use('default')
```

## Preprocessing
I create a smooth spiral-like path by letting `z` grow from 0 to 1 and using that value to shape `x` and `y`.


```python
z = np.linspace(0, 1, 100)
x = z * np.sin(25 * z)
y = z * np.cos(25 * z)
c = x + y

x[:5], y[:5], z[:5]
```

## 3D Line Plot
This view makes the full path easier to follow because depth is preserved.


```python
fig = plt.figure(figsize=(8, 6))
ax = plt.axes(projection='3d')
ax.plot3D(x, y, z, color='forestgreen', linewidth=2)
ax.set_title('3D Curve')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
plt.show()
```

## 3D Scatter Plot
Coloring the points by `x + y` adds a quick way to see how the values change along the curve.


```python
fig = plt.figure(figsize=(8, 6))
ax = plt.axes(projection='3d')
scatter = ax.scatter(x, y, z, c=c, cmap='viridis', s=40)
ax.set_title('3D Scatter View')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
fig.colorbar(scatter, ax=ax, label='x + y')
plt.show()
```

## 2D Projection
Dropping the `z` axis shows how much structure can disappear when depth is removed.


```python
plt.figure(figsize=(7, 5))
plt.scatter(x, y, c=z, cmap='plasma', s=35)
plt.title('2D Projection of the Same Curve')
plt.xlabel('x')
plt.ylabel('y')
plt.colorbar(label='z')
plt.show()
```

## What I Learned
- NumPy can generate useful synthetic signals quickly.
- A 3D plot can reveal structure that a 2D projection hides.
- Small notebooks like this are useful for practicing visualization before moving to real model outputs.
